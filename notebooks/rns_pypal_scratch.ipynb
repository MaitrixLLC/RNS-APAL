# RNS-PYPAL scratch notebook

Use this notebook for quick verification and interactive experiments while porting RNS-APAL to Python.

## 1. Locate the repository and import the library

This makes the notebook work whether the kernel starts in the repo root or inside the `notebooks` folder.

```python
from pathlib import Path
import sys

def find_repo_root(start=None):
    path = Path.cwd() if start is None else Path(start).resolve()
    for candidate in (path, *path.parents):
        if (candidate / "rns_pypal").exists() and (candidate / "tests").exists():
            return candidate
    raise RuntimeError("Could not find the RNS-PYPAL repository root")

repo_root = find_repo_root()
if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))

repo_root
```

```python
from rns_pypal import PPM, RNSNumberSystem

system = RNSNumberSystem(
    moduli=[3, 5, 7],
    powers=[1, 1, 1],
    fractional_digits=0,
    name="small-105",
)

system
```

## 2. Run the current test suite

This uses the same pytest configuration as the command line.

```python
from pathlib import Path
import sys
import pytest

def find_repo_root(start=None):
    path = Path.cwd() if start is None else Path(start).resolve()
    for candidate in (path, *path.parents):
        if (candidate / "rns_pypal").exists() and (candidate / "tests").exists():
            return candidate
    raise RuntimeError("Could not find the RNS-PYPAL repository root")

repo_root = find_repo_root()
if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))

result = pytest.main([str(repo_root / "tests")])
assert result == 0
```

## 3. Inspect simple PPM values

`format_native()` shows raw RNS residues. `format_whdr()` shows modulus headers above the residue row, matching the C++ `PrintDemo` style.

```python
a = PPM(10, system=system)
b = PPM(8, system=system)

print("a native:", a.format_native())
print("b native:", b.format_native())
print()
print(a.format_whdr())
```

```python
added = a.copy()
added.add(b)

multiplied = a.copy()
multiplied.mult(b)

print("a + b residues:", added.format_native())
print("a * b residues:", multiplied.format_native())
```

Reminder: these print routines display the residue digits. A full decimal-value print will come later through explicit conversion logic; arithmetic methods must not reconstruct values as Python integers to compute results.
