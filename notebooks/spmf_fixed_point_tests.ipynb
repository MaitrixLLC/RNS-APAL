```python
from pathlib import Path
import sys


def find_repo_root(start=None):
    path = Path.cwd() if start is None else Path(start).resolve()
    for candidate in (path, *path.parents):
        if (candidate / "rns_pypal").exists() and (candidate / "tests").exists():
            return candidate
    raise RuntimeError("Could not find the RNS-PYPAL repository root")


repo_root = find_repo_root()
if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))

from rns_pypal import (
    RNSNumberSystem,
    SPMF,
)

repo_root
```

## Fixed-point geometry

```python
from rns_pypal import SPMF, RNSNumberSystem

system = RNSNumberSystem(
    moduli=[2, 3, 5],
    powers=[4, 2, 1],
    fractional_digits=2,
    name="spmf-demo",
)

print("unit range:", system.unit_range)
print("fractional indices:", system.fractional_indices)
print("whole indices:", system.whole_indices)
```

## Assignment and representation

```python
x = SPMF(0, system=system)
x.assign_decimal("1.25")

print("native fixed:", x.format_fixed_native())
print("scaled int :", x.to_scaled_integer())
print("ratio      :", x.format_ratio())
print("decimal    :", x.format_decimal())
print("parts      :", x.fixed_parts())
print("dict       :", x.to_dict())
```

```python
y = SPMF(0, system=system)
y.assign_decimal("-0.5")

print("native fixed:", y.format_fixed_native())
print("scaled int :", y.to_scaled_integer())
print("ratio      :", y.format_ratio())
print("decimal    :", y.format_decimal())
print("parts      :", y.fixed_parts())
```

```python
w = SPMF(0, system=system)
w.assign_whole(2)

print("native fixed:", w.format_fixed_native())
print("scaled int :", w.to_scaled_integer())
print("ratio      :", w.format_ratio())
print("decimal    :", w.format_decimal())
```

```python
u_ppm = x.unit_ppm()
u_val = x.unit_value()

print("unit PPM native:", u_ppm.format_native())
print("unit PPM value :", u_ppm.format_value())

print("unit SPMF native:", u_val.format_fixed_native())
print("unit SPMF ratio :", u_val.format_ratio())
print("unit SPMF dec   :", u_val.format_decimal())
```

```python
z = SPMF(0, system=system)
z.assign_unit()

print("native fixed:", z.format_fixed_native())
print("scaled int :", z.to_scaled_integer())
print("ratio      :", z.format_ratio())
print("decimal    :", z.format_decimal())
```

## Arithmetic, comparison, and fraction position

```python
a = SPMF(0, system=system)
b = SPMF(0, system=system)

a.assign_decimal("1.25")
b.assign_decimal("0.5")

print("a before:", a.format_decimal(), a.format_ratio())
print("b       :", b.format_decimal(), b.format_ratio())

a.add(b)
print("a + b   :", a.format_decimal(), a.format_ratio(), a.format_fixed_native())

a.sub(b)
print("back    :", a.format_decimal(), a.format_ratio(), a.format_fixed_native())
```

```python
left = SPMF(0, system=system)
right = SPMF(0, system=system)

left.assign_decimal("1.25")
right.assign_decimal("0.5")

print(left.compare(right))   # 1 means left > right
print(right.compare(left))   # 0 means right is not > left
```

```python
v = SPMF(0, system=system)
v.assign_decimal("1.25")

print("normal:")
print("fractional:", v.fractional_indices)
print("whole     :", v.whole_indices)
print("unit range:", v.unit_range)
print("active?   :", v.sliding_point_active)

v.set_fraction_position(1)

print("\nslid:")
print("fractional:", v.fractional_indices)
print("whole     :", v.whole_indices)
print("unit range:", v.unit_range)
print("active?   :", v.sliding_point_active)

v.reset_fraction_position()

print("\nreset:")
print("fractional:", v.fractional_indices)
print("whole     :", v.whole_indices)
print("unit range:", v.unit_range)
print("active?   :", v.sliding_point_active)
```

```python
p = SPMF(0, system=system)
q = SPMF(0, system=system)

p.assign_decimal("1.0")
q.assign_decimal("0.5")
q.set_fraction_position(1)

try:
    p.add(q)
except Exception as exc:
    print(type(exc).__name__)
    print(exc)
```

```python
values = ["0", "0.25", "0.5", "1.0", "1.25", "-0.5"]

for text in values:
    value = SPMF(0, system=system)
    value.assign_decimal(text)
    print(
        text.rjust(6),
        "=> scaled:", str(value.to_scaled_integer()).rjust(5),
        "ratio:", value.format_ratio().rjust(8),
        "decimal:", value.format_decimal().rjust(8),
        "native:", value.format_fixed_native(),
    )
```

## Larger-system multiplication

```python
from rns_pypal import RNSNumberSystem, SPMF

system = RNSNumberSystem(
    moduli=[2, 3, 5, 7, 11, 13, 17],
    powers=[6, 4, 3, 2, 2, 1, 1],
    fractional_digits=3,
    name="larger-spmf-demo",
)

print("full moduli:", system.full_moduli)
print("dynamic range:", system.dynamic_range)
print("fractional unit range:", system.unit_range)
print("signed range:", SPMF.signed_range(system))
```

```python
x = SPMF(0, system=system)
y = SPMF(0, system=system)

x.assign_decimal("12.375")
y.assign_decimal("3.5")

print("x:")
x.print_fixed_native()
x.print_decimal()
x.print_ratio()

print("\ny:")
y.print_fixed_native()
y.print_decimal()
y.print_ratio()
```

```python
product = x.copy()
product.mult_std(y)

print("x * y:")
product.print_fixed_native()
product.print_decimal(max_fraction_digits=12)
product.print_ratio()

print("scaled integer:", product.to_scaled_integer())
print("sign flag:", product.sign_flag)
print("sign valid:", product.sign_valid)
```

```python
from rns_pypal import RNSNumberSystem, SPMF

system = RNSNumberSystem(
    moduli=[2, 3, 5, 7, 11, 13, 17, 19, 23],
    powers=[6, 4, 3, 3, 2, 2, 2, 1, 1],
    fractional_digits=3,
    name="larger-spmf-multiply-demo",
)

x = SPMF(0, system=system)
y = SPMF(0, system=system)

x.assign_decimal("12.375")
y.assign_decimal("3.5")

result = x.copy()
result.mult_std(y)

print("unit range:", system.unit_range)
print("dynamic range:", system.dynamic_range)
print("signed raw max:", (system.dynamic_range - 1) // 2)
print("result:", result.format_decimal(max_fraction_digits=12))
print("ratio:", result.format_ratio())
print("scaled integer:", result.to_scaled_integer())
```

```python
from rns_pypal import RNSNumberSystem, SPMF

system = RNSNumberSystem(
    moduli=[2, 3, 5, 7, 11, 13, 17, 19, 23],
    powers=[6, 4, 3, 3, 2, 2, 2, 1, 1],
    fractional_digits=3,
    name="signed-spmf-multiply-demo",
)

cases = [
    ("12.375", "3.5"),
    ("-12.375", "3.5"),
    ("12.375", "-3.5"),
    ("-12.375", "-3.5"),
]

for left_text, right_text in cases:
    left = SPMF(0, system=system)
    right = SPMF(0, system=system)

    left.assign_decimal(left_text)
    right.assign_decimal(right_text)

    product = left.copy()
    product.mult_std(right)

    print(f"{left_text} * {right_text}")
    print("  result decimal:", product.format_decimal(max_fraction_digits=12))
    print("  result ratio  :", product.format_ratio())
    print("  scaled integer:", product.to_scaled_integer())
    print("  sign flag     :", product.sign_flag)
    print("  sign valid    :", product.sign_valid)
    print("  native        :", product.format_fixed_native())
    print()
```

```python
for left_text, right_text in cases:
    left = SPMF(0, system=system)
    right = SPMF(0, system=system)

    left.assign_decimal(left_text)
    right.assign_decimal(right_text)

    product = left.copy()
    product.mult(right)

    print(
        f"{left_text:>9} × {right_text:>6} = "
        f"{product.format_decimal(max_fraction_digits=12):>12} "
        f"  ratio={product.format_ratio()} "
        f"  sign=({product.sign_flag}, valid={product.sign_valid})"
    )
```
