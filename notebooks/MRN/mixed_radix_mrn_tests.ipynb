# Mixed-Radix and MRN Tests

Focused experiments for streamed mixed-radix decomposition and materialized `MRN` conversion.

Automated verification: `tests/test_04_mixed_radix.py`. The numbered pytest files establish
the prerequisite hierarchy before this class or module.

```python
from pathlib import Path
import sys


def find_repo_root(start=None):
    path = Path.cwd() if start is None else Path(start).resolve()
    for candidate in (path, *path.parents):
        if (candidate / "rns_pypal").exists() and (candidate / "tests").exists():
            return candidate
    raise RuntimeError("Could not find the RNS-PYPAL repository root")


repo_root = find_repo_root()
if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))

from rns_pypal import (
    MRN,
    MixedRadixDecomposer,
    PPM,
    RNSNumberSystem,
    iter_mixed_radix_digits,
)

repo_root
```

## Stepwise mixed-radix decomposition

```python
system = RNSNumberSystem(
    moduli=[5, 7, 11],
    powers=[1, 1, 1],
)

x = PPM(26, system=system)

decomposer = MixedRadixDecomposer(x)

d0 = decomposer.step()
print(d0)
print(decomposer.working.format_native())
print(decomposer.working.format_whdr())

d1 = decomposer.step()
print(d1)
print(decomposer.working.format_native())
print(decomposer.is_complete)
```

## Materialized mixed-radix numbers

```python
x = PPM(384, system=system)
mrn = MRN.from_ppm(x)

print("MR digits, least-significant first:")
print([d.digit for d in mrn.digits])

print("MR radices:")
print([d.radix for d in mrn.digits])

print("MRN display, most-significant first:")
print(mrn.format_native())

print("Converted back to Python int for display/debug:")
print(mrn.to_int())
```
