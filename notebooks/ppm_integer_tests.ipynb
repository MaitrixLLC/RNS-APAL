```python
from rns_pypal import divide_residue_by_coprime_factor

value = 26

consumed_modulus = 5
remaining_modulus = 7

consumed_digit = value % consumed_modulus
remaining_residue = value % remaining_modulus

new_residue = divide_residue_by_coprime_factor(
    remaining_residue - consumed_digit,
    consumed_modulus,
    remaining_modulus,
)

print(consumed_digit)   # 1
print(new_residue)      # 5
```

```python
from rns_pypal import PPM, RNSNumberSystem

system = RNSNumberSystem(
    moduli=[2, 3, 5, 7, 11, 13],
    powers=[8, 5, 3, 2, 2, 1],
)

cases = [
    (123_456, 7),
    (987_654, 123),
    (12_345_678, 1_001),
    (system.dynamic_range - 1, 5),
    (system.dynamic_range - 1, 12_345),
]

for dividend_value, divisor_value in cases:
    quotient_holder = PPM(dividend_value, system=system)
    divisor = PPM(divisor_value, system=system)

    remainder = quotient_holder.div_std(divisor)

    q = int(quotient_holder.format_value())
    r = int(remainder.format_value())

    print(f"{dividend_value} / {divisor_value}")
    print("  quotient :", q)
    print("  remainder:", r)
    print("  expected :", dividend_value // divisor_value, dividend_value % divisor_value)
    print("  ok       :", q == dividend_value // divisor_value and r == dividend_value % divisor_value)
```

```python
from rns_pypal import RNSNumberSystem, SPMF

system = RNSNumberSystem(
    moduli=[2, 3, 5, 7, 11, 13, 17, 19, 23],
    powers=[6, 4, 3, 3, 2, 2, 2, 1, 1],
    fractional_digits=3,
    name="signed-spmf-multiply-demo",
)

cases = [
    ("12.375", "3.5"),
    ("-12.375", "3.5"),
    ("12.375", "-3.5"),
    ("-12.375", "-3.5"),
]

for left_text, right_text in cases:
    left = SPMF(0, system=system)
    right = SPMF(0, system=system)

    left.assign_decimal(left_text)
    right.assign_decimal(right_text)

    product = left.copy()
    product.mult_std(right)

    print(f"{left_text} * {right_text}")
    print("  result decimal:", product.format_decimal(max_fraction_digits=12))
    print("  result ratio  :", product.format_ratio())
    print("  scaled integer:", product.to_scaled_integer())
    print("  sign flag     :", product.sign_flag)
    print("  sign valid    :", product.sign_valid)
    print("  native        :", product.format_fixed_native())
    print()
```
