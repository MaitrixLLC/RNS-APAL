```python
from pathlib import Path
import sys


def find_repo_root(start=None):
    path = Path.cwd() if start is None else Path(start).resolve()
    for candidate in (path, *path.parents):
        if (candidate / "rns_pypal").exists() and (candidate / "tests").exists():
            return candidate
    raise RuntimeError("Could not find the RNS-PYPAL repository root")


repo_root = find_repo_root()
if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))

from rns_pypal import (
    PPM,
    RNSCriticalError,
    RNSNumberSystem,
)

repo_root
```

## Partial-power modular division

```python
system = RNSNumberSystem(
    moduli=[2, 3, 5],
    powers=[4, 2, 2],
)

x = PPM(42, system=system)

print("Before:")
print("value:", x.format_value())
print("raw:", x.format_native())
print(x.format_whdr())

x.mod_div(2)

print()
print("After x.mod_div(2):")
print("value:", x.format_value())
print("raw:", x.format_native())
print(x.format_whdr())

print()
for digit in x:
    print(
        f"index={digit.index}",
        f"base={digit.modulus}",
        f"power={digit.power}",
        f"power_valid={digit.power_valid}",
        f"skip={digit.skip}",
        f"digit={digit.digit}",
    )
```

```python
system = RNSNumberSystem(
    moduli=[2, 3, 5],
    powers=[4, 2, 2],
)

x = PPM(43, system=system)

print("Before:")
print("value:", x.format_value())
print("raw:", x.format_native())

try:
    x.mod_div(2)
except RNSCriticalError as error:
    print()
    print(error)

print()
print("After failed divide, should be unchanged:")
print("value:", x.format_value())
print("raw:", x.format_native())
print(x.format_whdr())
```

```python
system = RNSNumberSystem(
    moduli=[2, 3, 5],
    powers=[4, 2, 2],
)

derived = PPM(42, system=system)
derived.mod_div(2)

normalized = PPM(21, system=system)

print("derived raw:", derived.format_native())
print("normalized raw:", normalized.format_native())

try:
    derived.add(normalized)
except ValueError as error:
    print()
    print("Expected compatibility error:")
    print(error)
```

```python
system = RNSNumberSystem(
    moduli=[2, 3, 5],
    powers=[4, 2, 2],
)

a = PPM(42, system=system)
b = PPM(10, system=system)

a.mod_div(2)   # 21, derived format
b.mod_div(2)   # 5, same derived format

print("a:", a.format_value(), a.format_native())
print("b:", b.format_value(), b.format_native())

a.add(b)

print()
print("a + b:")
print("value:", a.format_value())
print("raw:", a.format_native())
print(a.format_whdr())
```

```python
system = RNSNumberSystem(
    moduli=[2, 3],
    powers=[1, 2],
)

x = PPM(6, system=system)

print("Original:")
print("value:", x.format_value())
print("raw:", x.format_native())
print(x.format_whdr())

x.mod_div(2)

print()
print("After mod_div(2), first digit skipped:")
print("value:", x.format_value())
print("raw:", x.format_native())
print(x.format_whdr())

x.normalize()

print()
print("After normalize(), skipped digit recovered:")
print("value:", x.format_value())
print("raw:", x.format_native())
print(x.format_whdr())
```

```python
system = RNSNumberSystem(
    moduli=[2, 3, 5],
    powers=[4, 2, 2],
)

derived = PPM(42, system=system)
derived.mod_div(2)

normalized = derived.normalized_copy()

print("Derived source remains:")
print("raw:", derived.format_native())
print(derived.format_whdr())

print()
print("Normalized copy:")
print("raw:", normalized.format_native())
print(normalized.format_whdr())
```

## Standard integer division

```python
from rns_pypal import PPM, RNSNumberSystem

system = RNSNumberSystem(
    moduli=[2, 3, 5, 7],
    powers=[4, 2, 1, 1],
)

dividend = PPM(123, system=system)
divisor = PPM(10, system=system)

remainder = dividend.div_std(divisor)

print("quotient:")
dividend.print_whdr()
print("quotient value:", dividend.format_value())

print("remainder:")
remainder.print_whdr()
print("remainder value:", remainder.format_value())
```

```python
from rns_pypal import PPM, RNSNumberSystem

system = RNSNumberSystem(
    moduli=[2, 3, 5, 7, 11, 13],
    powers=[8, 5, 3, 2, 2, 1],
    name="larger_ppm_test",
)

print("dynamic range:", system.dynamic_range)

dividend_value = 123_456_789
divisor_value = 12_345

quotient_holder = PPM(dividend_value, system=system)
divisor = PPM(divisor_value, system=system)

remainder = quotient_holder.div_std(divisor)

print("quotient:")
quotient_holder.print_whdr()
print("quotient value:", quotient_holder.format_value())

print("\nremainder:")
remainder.print_whdr()
print("remainder value:", remainder.format_value())

print("\nPython check:")
print("expected quotient:", dividend_value // divisor_value)
print("expected remainder:", dividend_value % divisor_value)
```

```python
from rns_pypal import PPM, RNSNumberSystem

system = RNSNumberSystem(
    moduli=[2, 3, 5, 7, 11, 13],
    powers=[10, 6, 4, 3, 2, 2],
    name="high_range_ppm_test",
)

largest = system.dynamic_range - 1
divisor_value = 97_531

quotient_holder = PPM(largest, system=system)
divisor = PPM(divisor_value, system=system)

remainder = quotient_holder.div_std(divisor)

print("dynamic range:", system.dynamic_range)
print("dividend:", largest)
print("divisor:", divisor_value)

print("\nquotient value:", quotient_holder.format_value())
print("remainder value:", remainder.format_value())

print("\nPython check:")
print(largest // divisor_value)
print(largest % divisor_value)
```

```python
from rns_pypal import PPM, RNSNumberSystem

system = RNSNumberSystem(
    moduli=[2, 3, 5, 7, 11, 13],
    powers=[8, 5, 3, 2, 2, 1],
)

cases = [
    (123_456, 7),
    (987_654, 123),
    (12_345_678, 1_001),
    (system.dynamic_range - 1, 5),
    (system.dynamic_range - 1, 12_345),
]

for dividend_value, divisor_value in cases:
    quotient_holder = PPM(dividend_value, system=system)
    divisor = PPM(divisor_value, system=system)

    remainder = quotient_holder.div_std(divisor)

    q = int(quotient_holder.format_value())
    r = int(remainder.format_value())

    print(f"{dividend_value} / {divisor_value}")
    print("  quotient :", q)
    print("  remainder:", r)
    print("  expected :", dividend_value // divisor_value, dividend_value % divisor_value)
    print("  ok       :", q == dividend_value // divisor_value and r == dividend_value % divisor_value)
```

## Nested PPM collections

```python
from rns_pypal import PPM, RNSNumberSystem

system = RNSNumberSystem(
    moduli=[2, 3, 5, 7],
    powers=[4, 2, 1, 1],
    name="matrix-demo",
)

matrix = [
    [PPM(1, system=system), PPM(2, system=system), PPM(3, system=system)],
    [PPM(4, system=system), PPM(5, system=system), PPM(6, system=system)],
]

for row in matrix:
    print([x.format_value() for x in row])
```

```python
for row in matrix:
    for x in row:
        x.mult(3)
        x.add(1)

for row in matrix:
    print([x.format_value() for x in row])
```
